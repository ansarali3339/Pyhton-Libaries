```python
plt.figure(figsize=(6,5),facecolor="#E8FEB8DA")
a=sns.boxplot(x='sex',y='age',data=ti)
plt.gca().set_facecolor("#A7EA8EDA")
```

```python
# To remove the outlier:
#  formula:  IQR=Q3-Q1  [ Inter quartile range ]

#IQR
# q3= var['col_name'].quantile(0.75)
# q1= var['col_name'].quantile(0.25)
# lower_bound=q1-0.5*IQR
# upper_bound=q3+0.5*IQR
```

```python
q3= ti['age'].quantile(0.75)
q1= ti['age'].quantile(0.25)
IQR=q3-q1 
IQR
```

```python
ti['age'].describe()
lower_bound=q1-0.5*IQR
upper_bound=q3+0.5*IQR
lower_bound,upper_bound
```

```python
outlier=ti[(ti['age']>upper_bound) | (ti['age']<lower_bound)]
outlier
```

```python
#how to handle outliers
out=[]
for i in ti['age']:
    if i > upper_bound:
        out.append(upper_bound)
    elif i < lower_bound:
        out.append(lower_bound)
    else:
        out.append(i)
ti['age']=out
```

```python
plt.figure(figsize=(6,5),facecolor="#E8FEB8DA")
a=sns.boxplot(x='sex',y='age',data=ti)
plt.gca().set_facecolor("#A7EA8EDA")
```
