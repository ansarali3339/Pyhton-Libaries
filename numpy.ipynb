```python
import numpy as np
# aggregate functions
#1) np.sum(var) --> it will return the sum of all the values present in the array
#2) np.min(var) --> it will return the min value present in the array
#3) np.max(var) --> it will return the max value present in the array
#4)np.mean(var) --> it will return mean or avg value of the array

a=np.arange(1,25)
b=a.reshape((4,6))
c=a.reshape((3,2,4))

np.sum(a)
np.sum(b)
np.sum(c)


np.min(a)
np.min(b)
np.min(c)


np.max(a)
np.max(b)
np.max(c)

print(a.ndim)
print(b.ndim)
print(c.ndim)

np.mean(a)
b.mean()
c.mean()
```

```python
#Filtering conditions
#syntax :- var[condition]


ages=np.array([[17,34,65,23,72,18,45,12,25,11],
               [22,42,16,46,38,12,62,54,16,28]])

#FIND AGE GREATER THAN 60
ages[ages>60]

#find age <18
ages[ages<18]

a=np.arange(1,25)
a
a[a>18]


#logical operators
#AND(&)
#OR (|)
#NOT (~)
 

 #find ages which are >18 and <60

ages=np.array([[17,34,65,23,72,18,45,12,25,11],
               [22,42,16,46,38,12,62,54,16,28]])

ages[(ages>18) & (ages<60)]
(ages>18) & (ages<60) 
```

```python
a=np.arange(1,25)

a[( a % 5 ==0 ) | (a % 7== 0)]
```

```python

ages[~((ages<18) | (ages>60))]
```

```python
#find age>18 and age<60
ages[(ages<18) | (ages>60)]

# or

ages[~((ages>18) & (ages<60))]

```
