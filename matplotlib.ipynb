```python
#Data Visualization
#     It is a process of converting the data into pictorial or graphical representation to understand the data in a better manner.
# three ways of data visualization are:
# 1)matplotlib
# 2)Seaborn(lib)
# 3)Power BI


import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
#plot():
#    It is used to create 2D graph
# Syntax:
#      plt.plot(x,y)

#plt.figure(figsize=(h,w))  it is used to adjust the height and width of the chart
#va='top' va stands for vertical alignment
#ha='right' ha stands for horizontal alignment
```

```python
x=np.arange(1,11)
y=x**2
z=x**3
```

```python
plt.figure(figsize=(12,5))  #we have to use figure in 1st line 
plt.plot(x,y,label='X-values',marker='h',markersize=8,color='black',markerfacecolor='pink',markeredgecolor='red',linewidth=1,linestyle='dashed')


plt.legend(title='values',fontsize=15,loc='upper center')
plt.xlabel('value 1 to 10',fontsize=15)
plt.ylabel('square value',fontsize=15)
plt.title('square of numbers',fontsize=18,color='black')

#Annotate
for i in range(len(x)):  #this is use to get the index value of x
    plt.annotate(f'({x[i]},{y[i]})', (x[i],y[i]),va='top',ha='left')   #formatted string or f '' is use to access the variable in the print only by using {var}these brackets. the second x and y value are use to get where to print.
```

```python
plt.figure(figsize=(12,5))  
plt.plot(x,z,label='X-values',marker='h',markersize=8,color='black',markerfacecolor='pink',markeredgecolor='red',linewidth=1,linestyle='dashed')


plt.legend(title='values',fontsize=15,loc='upper center')
plt.xlabel('value 1 to 10',fontsize=15)
plt.ylabel('cube value',fontsize=15)
plt.title('cube of numbers',fontsize=18,color='black')
plt.grid(True)

#Annotate
for i in range(len(x)):  #this is use to get the index value of x  
    plt.annotate(f'({x[i]},{z[i]})', (x[i],z[i]),va='top',ha='left')

```
